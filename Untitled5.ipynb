```python
import PyPDF2

def extract_text_from_pdf(pdf_path):
    text = ""

    with open(pdf_path, "rb") as pdf:
        reader = PyPDF2.PdfReader(pdf)

        for page in reader.pages:
            text += page.extract_text() or ""

    return text


physics_text = extract_text_from_pdf("Physics_Force.pdf")
chemistry_text = extract_text_from_pdf("Chemistry_Matter.pdf")
```

```python
import pytesseract
from pdf2image import convert_from_path

if len(chemistry_text.strip()) < 100:

    print("Chemistry PDF looks image-based. Using OCR...")

    pages = convert_from_path("Chemistry_Matter.pdf")

    chemistry_text = ""

    for page_image in pages:
        chemistry_text += pytesseract.image_to_string(page_image)

    print(
        "OCR done. New Chemistry text length:",
        len(chemistry_text)
    )

else:
    print("Chemistry PDF already contains readable text.")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# chunk_size = characters per chunk
# chunk_overlap = shared characters between chunks
# This helps preserve context between chunks

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

physics_chunks = splitter.split_text(physics_text)
chemistry_chunks = splitter.split_text(chemistry_text)

print("Physics PDF:", len(physics_chunks), "chunks")
print("Chemistry PDF:", len(chemistry_chunks), "chunks")
```

```python
import chromadb
from sentence_transformers import SentenceTransformer

# Load embedding model
# Converts text into numerical vectors

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# Create persistent ChromaDB client
# The database will be stored in the "chroma_db" folder

client = chromadb.PersistentClient(
    path="./chroma_db"
)
```

```python
# Create vector collection

collection = client.get_or_create_collection(
    name="pdf_docs",
    metadata={
        "hnsw:space": "cosine"
    }
)

print("Collection created successfully.")
```

```python
def add_chunks_to_db(chunks, source_name):

    embeddings = model.encode(
        chunks
    ).tolist()

    ids = [
        f"{source_name}_{i}"
        for i in range(len(chunks))
    ]

    metadatas = [
        {
            "source": source_name
        }
        for _ in chunks
    ]

    collection.add(
        documents=chunks,
        embeddings=embeddings,
        ids=ids,
        metadatas=metadatas
    )
```

```python
add_chunks_to_db(
    physics_chunks,
    "Physics_Force"
)

add_chunks_to_db(
    chemistry_chunks,
    "Chemistry_Matter"
)

print(
    "Total items in database:",
    collection.count()
)
```

```python
sample = collection.peek(limit=5)

for doc, meta in zip(
    sample["documents"],
    sample["metadatas"]
):
    print(
        f"[{meta['source']}] {doc[:200]}..."
    )
    print()
```

```python
def search(query, n_results=3, source_filter=None):

    # Convert the user's question into an embedding
    query_embedding = model.encode(query).tolist()

    # Search parameters
    kwargs = {
        "query_embeddings": [query_embedding],
        "n_results": n_results
    }

    # Optional source filter
    if source_filter:
        kwargs["where"] = {
            "source": source_filter
        }

    # Search ChromaDB
    results = collection.query(**kwargs)

    # Display results
    for i in range(len(results["documents"][0])):

        distance = results["distances"][0][i]

        # With cosine distance:
        # similarity = 1 - distance

        similarity = 1 - distance

        source = results["metadatas"][0][i]["source"]

        print(
            f"--- Result {i+1} | "
            f"Source: {source} | "
            f"Similarity: {similarity:.3f} ---"
        )

        print(results["documents"][0][i])
        print()
```

```python
print("PHYSICS QUESTIONS")
print("=" * 60)

search(
    "What is force and what are its effects?"
)

search(
    "What is Newton's first law of motion?"
)

search(
    "What is Newton's second law of motion?"
)

search(
    "What is the relationship between force mass and acceleration?"
)
```

```python
print("CHEMISTRY QUESTIONS")
print("=" * 60)

search(
    "What is matter?"
)

search(
    "What are the different states of matter?"
)

search(
    "What is a solid?"
)

search(
    "What is a liquid?"
)

search(
    "What is a gas?"
)
```
